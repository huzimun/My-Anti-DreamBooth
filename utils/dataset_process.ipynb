```python
import os
import shutil

# 指定根目录
root_dir = "/data1/humw/Datasets/new-CelebA-HQ"

# 遍历根目录下的所有子目录
for subdir in os.listdir(root_dir):
    subdir_path = os.path.join(root_dir, subdir)
    
    # 检查是否为目录
    if os.path.isdir(subdir_path):
        # 在子目录下创建set_B目录
        set_b_dir = os.path.join(subdir_path, "set_B")
        os.makedirs(set_b_dir, exist_ok=True)
        
        # 读取子目录下所有文件的路径
        for filename in os.listdir(subdir_path):
            file_path = os.path.join(subdir_path, filename)
            
            # 检查是否为图像文件（可以根据需要扩展文件类型）
            if filename.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp', '.gif')):
                # 将图像文件移动到set_B目录下
                shutil.move(file_path, os.path.join(set_b_dir, filename))

print("处理完成！")
```

```python
# 构造对抗扰动数据集，将对抗样本数据集的每张图像替换为对抗扰动
def replace_with_perturbations(original_dir, perturbation_dir):
    """
    将原始图像目录中的图像替换为对应的对抗扰动图像。
    
    :param original_dir: 原始图像目录
    :param perturbation_dir: 对抗扰动图像目录
    """
    # 遍历原始图像目录
    for filename in os.listdir(original_dir):
        original_path = os.path.join(original_dir, filename)
        perturbation_path = os.path.join(perturbation_dir, filename)
        
        # 检查原始图像是否存在对应的对抗扰动图像
        if os.path.exists(perturbation_path):
            # 替换原始图像为对抗扰动图像
            shutil.move(perturbation_path, original_path)
        else:
            print(f"Warning: 对抗扰动图像 {perturbation_path} 不存在，无法替换 {original_path}")
```
