```python
import os

ted_path_list = list()
other_path_list = list()

for person_id in os.listdir(ted_yingbu_dir):
    yingbu_dir = os.path.join(ted_yingbu_dir, person_id)
    lvmeng_dir = os.path.join(ted_lvmeng_dir, person_id)
    chengyaojin_dir = os.path.join(ted_chengyaojin_dir, person_id)
    caocao_dir = os.path.join(ted_caocao_dir, person_id)
    guanyu_dir = os.path.join(ted_guanyu_dir, person_id)
    for image_name in os.listdir(caocao_dir):
        image_path = os.path.join(caocao_dir, image_name)
        ted_path_list.append(image_path)
    for image_name in os.listdir(chengyaojin_dir):
        image_path = os.path.join(chengyaojin_dir, image_name)
        ted_path_list.append(image_path)
    for image_name in os.listdir(lvmeng_dir):
        image_path = os.path.join(lvmeng_dir, image_name)
        ted_path_list.append(image_path)
    for image_name in os.listdir(yingbu_dir):
        image_path = os.path.join(yingbu_dir, image_name)
        ted_path_list.append(image_path)
    for image_name in os.listdir(guanyu_dir):
        image_path = os.path.join(guanyu_dir, image_name)
        ted_path_list.append(image_path)

for person_id in os.listdir(other_idprotector_dir):
    idprotector_dir = os.path.join(other_idprotector_dir, person_id)
    vae_dir = os.path.join(other_ace_dir, person_id)
    ipadapter_dir = os.path.join(other_anti_db_dir, person_id)
    photomaker_dir = os.path.join(other_caat_dir, person_id)
    pap_dir = os.path.join(other_pap_dir, person_id)
    for image_name in os.listdir(idprotector_dir):
        image_path = os.path.join(idprotector_dir, image_name)
        other_path_list.append(image_path)
    for image_name in os.listdir(vae_dir):
        image_path = os.path.join(vae_dir, image_name)
        other_path_list.append(image_path)
    for image_name in os.listdir(ipadapter_dir):
        image_path = os.path.join(ipadapter_dir, image_name)
        other_path_list.append(image_path)
    for image_name in os.listdir(photomaker_dir):
        image_path = os.path.join(photomaker_dir, image_name)
        other_path_list.append(image_path)
    for image_name in os.listdir(pap_dir):
        image_path = os.path.join(pap_dir, image_name)
        other_path_list.append(image_path)
```

```python
# 复制图像到对应的目录下
import shutil
def copy_images(image_list, target_dir):
    for i in range(0, len(image_list)):
        image_path = image_list[i]
        # 设置目标图像名为i+原始图像名
        target_path = os.path.join(target_dir, str(i) + "_" + image_path[image_path.rfind("/")+1:])
        shutil.copy(image_path, target_path)

train_ted_dir = "/data1/humw/Codes/My-Anti-DreamBooth/data/perturbed_images_dataset_v5/train/ted"
copy_images(train_ted_path_list, train_ted_dir)
train_other_dir = "/data1/humw/Codes/My-Anti-DreamBooth/data/perturbed_images_dataset_v5/train/other"
copy_images(train_other_path_list, train_other_dir)
val_ted_dir = "/data1/humw/Codes/My-Anti-DreamBooth/data/perturbed_images_dataset_v5/val/ted"
copy_images(val_ted_path_list, val_ted_dir)
val_other_dir = "/data1/humw/Codes/My-Anti-DreamBooth/data/perturbed_images_dataset_v5/val/other"
copy_images(val_other_path_list, val_other_dir)
```
